```python
import numpy as np 
import pandas as pd
import seaborn as sns
```

```python
df = sns.load_dataset("diamonds")
df.head()
```

```python
## chehcking null values
df.isnull().sum()
```

```python
### seprating data as x and y
y = df[["price"]]
x = df.drop(labels="price",axis=1)
```

```python
x.info()
```

```python
x_num = x.select_dtypes(include="float64")
x_num.head()
```

```python
x_char = x.select_dtypes(exclude="float64")
from sklearn.preprocessing import LabelEncoder
import pickle

# Make independent copy
x_char = x.select_dtypes(exclude="float64").copy()

# Encode categorical columns ONCE
cut_le = LabelEncoder()
x_char["cut"] = cut_le.fit_transform(x_char["cut"])

color_le = LabelEncoder()
x_char["color"] = color_le.fit_transform(x_char["color"])

clarity_le = LabelEncoder()
x_char["clarity"] = clarity_le.fit_transform(x_char["clarity"])

# Save the encoders
with open("cut_le.pkl", "wb") as f:
    pickle.dump(cut_le, f)

with open("color_le.pkl", "wb") as f:
    pickle.dump(color_le, f)

with open("clarity_le.pkl", "wb") as f:
    pickle.dump(clarity_le, f)

# Check
print("CUT:", cut_le.classes_)
print("COLOR:", color_le.classes_)
print("CLARITY:", clarity_le.classes_)
```

```python

pickle.dump(cut_le, open("cut_le.pkl", "wb"))
pickle.dump(color_le, open("color_le.pkl", "wb"))
pickle.dump(clarity_le, open("clarity_le.pkl", "wb"))
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
```

```python
x_scaling = pd.DataFrame(sc.fit_transform(x_num),
                         columns = x_num.columns,
                         index=x_num.index)
x_scaling.head()
```

```python
data = pd.concat([x_char,x_scaling],axis=1)
data.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(data,y,
                                                test_size=0.2,
                                                random_state=45)
```

```python
from sklearn.neighbors import KNeighborsRegressor
knn =KNeighborsRegressor(n_neighbors=4)
```

```python
knn.fit(x_train,y_train)
```

```python
ypred=knn.predict(x_test)
from sklearn.metrics import r2_score
print(r2_score(y_test,ypred))
```

```python
pickle.dump(knn,open("diamond_knn.pkl","wb"))
pickle.dump(sc,open("diamond_scaling.pkl","wb"))
```
